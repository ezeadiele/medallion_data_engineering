#Read Data from csv and write into the bronze iteratively

```python
# Path to the source folder in your Volume
source_path = "/Volumes/workspace/bronze/source_systems/source_crm"

# List all files in that folder
files = dbutils.fs.ls(source_path)

for file in files:
    file_name = file.name  # e.g. "cust_info.csv"

    # Only process CSV files
    if file_name.endswith(".csv"):
        table_name = file_name.replace(".csv", "")  # e.g. "cust_info"

        print(f"Reading {file_name} -> writing to table bronze.{table_name}")

        df = (
            spark.read.option("header", "True")
            .option("inferSchema", "True")
            .csv(file.path)
        )

        # Write to a bronze Delta table named after the file
        df.write.mode("overwrite").saveAsTable(f"workspace.bronze.{table_name}")
```
